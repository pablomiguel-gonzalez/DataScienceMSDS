# **EDA**

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel('DirtyData.xlsx') ## we load the dirty data
print(df)
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True) ## we follow the KNIME workflow, we start by removing duplicates
print(df)
```

```python
df['First name'] = df['First name'].str.strip() ## we remove the spaces
df['First name'] = df['First name'].str.title() ## we capitalize only the first letter of each word
print(df)
```

```python
df['Surname'] = df['Surname'].str.strip() ## we remove the spaces, now for Surnames
df['Surname'] = df['Surname'].str.title() ## same thing, for surnames
print(df)
```

```python
df['Full name'] = df['First name'] + " " + df['Surname'] ## we create a new column with the full name
df.drop(['First name', 'Surname'], axis=1, inplace=True) ## we drop the old columns
print(df)
```

```python
df = df[df['__Email_Entered'].str.contains(r'^[\w\.-]+@[\w\.-]+\.\w+$', na = False, regex = True)] ## we remove the invalid emails
print(df)
```

```python
df = df[df['_CREDIT'] >= 0] ## we remove the negative credits
print(df)
```

```python
df['Logins'] = df['Logins'].fillna(0) ## we fill the missing values with 0
df['Age'] = df['Age'].fillna(df['Age'].median()) ## we fill the missing values with the mean
print(df)
```

```python
df.drop('IP_Address', axis=1, inplace=True) ## we drop the uneeded column
print(df)
```

```python
df.rename(columns={
    '__Email_Entered': 'Email',  ## we improve the column names
    '_CREDIT': 'Credit'
}, inplace=True)
print(df)
```

```python
df = df[['Full name', 'Email', 'Gender', 'Age','Logins', 'Credit']] ## adjust column order
print(df)
```

```python
df.to_excel('cleaned_data.xlsx', index=False)
## save cleaned dataset
```

# **Visualizations**

```python
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Age', y='Logins')
plt.title('Age vs. Logins')
plt.xlabel('Age')
plt.ylabel('Logins')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Age', y='Credit')
plt.title('Age vs. Credit Score')
plt.xlabel('Age')
plt.ylabel('Credit Score')
plt.tight_layout()
plt.show()
```

```python
gender_counts = df['Gender'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Gender Distribution')
plt.axis('equal')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Age', bins=20)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

# **Binning**

```python
df['Age_bin'] = pd.cut(df['Age'], bins=[0, 30, 45, 60, 100], labels=["<30", "30–45", "45–60", "60+"])
df['Credit_bin'] = pd.cut(df['Credit'], bins=[0, 50, 100, 150], labels=["Poor", "Fair", "Good"])
```

```python
sns.countplot(x='Age_bin', data=df)
plt.title('Customers by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.show()
```

```python
sns.countplot(x='Credit_bin', data=df)
plt.title('Customers by Credit Score Category')
plt.xlabel('Credit Bin')
plt.ylabel('Count')
plt.show()
```
